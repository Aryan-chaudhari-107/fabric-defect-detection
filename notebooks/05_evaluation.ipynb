## 2. Load Tuned Thresholds

English: Optimal patch threshold and minimum patch count tuned strictly on validation images.

Hindi: Validation set par grid search se nikale gaye best decision thresholds.

````python
for model_name in ["cnn", "efficientnet"]:
    thresh_path = METRICS_DIR / f"stage1_thresholds_{model_name}.json"
    if thresh_path.exists():
        with thresh_path.open("r", encoding="utf-8") as f:
            data = json.load(f)
        print(f"\n=== {model_name.upper()} Tuned Thresholds ===")
        print(f"P(defective) threshold : {data['best_stage1_prob_threshold']}")
        print(f"Min defective patches  : {data['best_min_defective_patches']}")
        print(f"Val Balanced Accuracy  : {data['val_balanced_accuracy']:.4f}")
        print(f"Val Defect Recall (TPR): {data['val_defect_recall']:.4f}")
        print(f"Val Normal FPR         : {data['val_normal_false_positive_rate']:.4f}")
    else:
        print(f"No threshold file found for {model_name}. Run `python src/tune_thresholds.py --model {model_name}` first.")
````
